```python
# 1. Normalize & Hash Set (Fastest, O(n))

# For each triple (h, r, t), create a canonical form: (min(h, t), max(h, t), r)
# Use a set to track seen normalized triples
# Keep only the first occurrence of each canonical form
# Skip when you encounter the normalized pair again

# 2. Bidirectional Lookup Index (O(n) with parallel processing)

# Build a map: {(min(h,t), max(h,t), r) → [list of original triples]}
# For each canonical form with multiple triples, keep one and flag others for deletion
# Good if you need to know which inverse pair corresponds to which triple

# 3. Sort-based Grouping (O(n log n), good for huge KGs)

# Sort all triples by (min(h, t), max(h, t), r) as composite key
# Iterate once and skip consecutive duplicates
# Memory efficient, scales well to very large graphs

# 4. Vectorized NumPy Approach (Fastest for massive datasets)

# Load triples as structured NumPy array with fields: head, tail, relation
# Create canonical column: (np.minimum(h, t), np.maximum(h, t), r)
# Use np.unique() with axis or fancy indexing to deduplicate in one pass
# Can handle millions of triples efficiently
```

```python
def filter_implicit_inverses(df, h_name="x_name", t_name="y_name", 
                             h_type="x_type", t_type="y_type", rel="relation"):
    """
    Filters out implicit inverse triples by enforcing a deterministic direction.
    Rule 1: Sort by node type alphabetically.
    Rule 2: If types are the same, sort by node name alphabetically.
    """
    
    # Extract as numpy arrays for speed
    hn = df[h_name].astype(str).values
    tn = df[t_name].astype(str).values
    ht = df[h_type].astype(str).values
    tt = df[t_type].astype(str).values
    
    # Boolean mask: True if the current edge is ALREADY in the correct canonical order
    # (head type < tail type) OR (types are equal AND head name <= tail name)
    is_canonical = (ht < tt) | ((ht == tt) & (hn <= tn))
    
    # Construct the canonical head, tail, and types using the mask
    canon_head = np.where(is_canonical, hn, tn)
    canon_tail = np.where(is_canonical, tn, hn)
    canon_htype = np.where(is_canonical, ht, tt)
    canon_ttype = np.where(is_canonical, tt, ht)
    
    # Create a temporary dataframe to run deduplication
    temp = pd.DataFrame({
        'canon_head': canon_head,
        'canon_tail': canon_tail,
        'canon_htype': canon_htype,
        'canon_ttype': canon_ttype,
        'relation': df[rel].values,
        '_original_idx': df.index.values
    })
    
    # Drop duplicates on the canonical representation
    # 'keep="first"' ensures we just keep one of the edges (effectively deleting the reverse)
    dedup_idx = temp.drop_duplicates(
        subset=['canon_head', 'canon_tail', 'relation'], 
        keep='first'
    )['_original_idx'].values
    
    # Return the original dataframe filtered down to the remaining indices
    return df.loc[dedup_idx].reset_index(drop=True)
```

```python
def remove_inverse_relations(
    df: DataFrame,
    head_col: str = "x_name",
    tail_col: str = "y_name",
    relation_col: str = "relation",
) -> DataFrame:
    out = df.copy()
    left = out[head_col].astype(str)
    right = out[tail_col].astype(str)

    canonical_left = left.where(left <= right, right)
    canonical_right = right.where(left <= right, left)

    out = out.assign(_canonical_left=canonical_left, _canonical_right=canonical_right)
    out = out.drop_duplicates(subset=["_canonical_left", "_canonical_right", relation_col])
    return out.drop(columns=["_canonical_left", "_canonical_right"]).reset_index(drop=True)
```

```python
# === Vectorized set-based dedupe (fast!) ===
def remove_inverse_relations_fast(df, head_col="x_name", tail_col="y_name", relation_col="relation"):
    """
    Vectorized removal of inverse triples.
    Treats (h, r, t) and (t, r, h) as identical by canonical ordering.
    ~1000x faster than iterrows approach.
    """
    # Extract columns as numpy arrays
    heads = df[head_col].astype(str).values
    tails = df[tail_col].astype(str).values
    rels = df[relation_col].astype(str).values
    
    # Vectorized: compute canonical form for all rows at once
    canonical_left = np.where(heads <= tails, heads, tails)
    canonical_right = np.where(heads <= tails, tails, heads)
    
    # Create a temporary dataframe for deduplication
    temp = pd.DataFrame({
        'canonical_left': canonical_left,
        'canonical_right': canonical_right,
        'relation': rels,
        '_idx': range(len(df))
    })
    
    # Keep only first occurrence of each canonical triple
    dedup_idx = temp.drop_duplicates(subset=['canonical_left', 'canonical_right', 'relation'], keep='first')['_idx'].values
    
    return df.iloc[dedup_idx].reset_index(drop=True)

# Test it
print(f"Original shape: {deduped_prime_kg_df.shape}")
import time
start = time.time()
test_set_dedup_fast = remove_inverse_relations_fast(deduped_prime_kg_df)
elapsed = time.time() - start
print(f"After vectorized set-based dedup: {test_set_dedup_fast.shape}")
print(f"Rows removed: {deduped_prime_kg_df.shape[0] - test_set_dedup_fast.shape[0]}")
print(f"Time: {elapsed:.2f}s")

# Compare with dataframe-based approach
start = time.time()
test_df_dedup = remove_inverse_relations(deduped_prime_kg_df)
elapsed_df = time.time() - start
print(f"\nDataframe-based dedup: {test_df_dedup.shape}")
print(f"Time: {elapsed_df:.2f}s")
print(f"Speedup: {elapsed_df / elapsed:.1f}x")
print(f"Same result? {test_set_dedup_fast.shape == test_df_dedup.shape}")
```

```python
EXPLICIT_DIRECTIONS = {
    "anatomy_protein_present":    ("anatomy", "gene/protein"),
    "disease_phenotype_positive": ("disease", "effect/phenotype"),
    "bioprocess_protein":         ("biological_process", "gene/protein"),
    "cellcomp_protein":           ("cellular_component", "gene/protein"),
    "disease_protein":            ("disease", "gene/protein"),
    "molfunc_protein":            ("molecular_function", "gene/protein"),
    "drug_effect":                ("drug", "effect/phenotype"),
    "pathway_protein":            ("pathway", "gene/protein"),
    "contraindication":           ("drug", "disease"),
    "drug_protein":               ("drug", "gene/protein"),
    "anatomy_protein_absent":     ("anatomy", "gene/protein"),
    "indication":                 ("drug", "disease"),
    "phenotype_protein":          ("effect/phenotype", "gene/protein"),
    "off-label use":              ("drug", "disease"),
    "exposure_disease":           ("exposure", "disease"),
    "exposure_bioprocess":        ("exposure", "biological_process"),
    "exposure_protein":           ("exposure", "gene/protein"),
    "disease_phenotype_negative": ("disease", "effect/phenotype"),
    "exposure_molfunc":           ("exposure", "molecular_function"),
    "exposure_cellcomp":          ("exposure", "cellular_component"),
}


# DOESENT FLIP THE OTHER VALUES JUST THE NAME!!!
def enforce_explicit_directions(df, target_mapping, h_name="x_name", t_name="y_name", 
                                h_type="x_type", t_type="y_type", rel="relation"):
    df_out = df.copy()
    
    hn = df_out[h_name].astype(str).values
    tn = df_out[t_name].astype(str).values
    ht = df_out[h_type].astype(str).values
    tt = df_out[t_type].astype(str).values
    rels = df_out[rel].astype(str).values
    
    # We will build a boolean mask of rows that are "backwards" and need to be flipped
    swap_mask = np.zeros(len(df_out), dtype=bool)
    
    # 2. Force Explicit Rule Directions
    for relation, (req_head_type, req_tail_type) in target_mapping.items():
        rel_mask = (rels == relation)
        # If the row is this relation, but the head type matches the required TAIL type, it's backwards!
        needs_swap = rel_mask & (ht == req_tail_type) & (tt == req_head_type)
        swap_mask = swap_mask | needs_swap

    # 3. Handle Same-Type Relations (e.g., drug_drug, disease_disease)
    # For these, we just enforce alphabetical order of the node names so they can be deduplicated.
    same_type_mask = (ht == tt)
    alpha_swap = same_type_mask & (hn > tn)
    swap_mask = swap_mask | alpha_swap
    
    # Optional 4: Fallback for entirely unlisted relations connecting different types
    # (Alphabetical by node type just to guarantee deduplication works)
    unlisted_mask = ~(same_type_mask | np.isin(rels, list(target_mapping.keys())))
    fallback_swap = unlisted_mask & (ht > tt)
    swap_mask = swap_mask | fallback_swap
    
    # 5. Actively OVERWRITE the dataframe arrays with the flipped values
    df_out[h_name] = np.where(swap_mask, tn, hn)
    df_out[t_name] = np.where(swap_mask, hn, tn)
    df_out[h_type] = np.where(swap_mask, tt, ht)
    df_out[t_type] = np.where(swap_mask, ht, tt)
    
    # 6. Now that every inverse has been physically flipped to match the exact same string...
    # drop_duplicates will perfectly swallow the inverses and keep exactly 1 copy.
    df_out = df_out.drop_duplicates(subset=[h_name, rel, t_name]).reset_index(drop=True)
    
    return df_out

# Execute:
final_kg = enforce_explicit_directions(prime_kg_df, target_mapping=EXPLICIT_DIRECTIONS)
relation_type_stats(final_kg)
```

```python
def enforce_explicit_directions_safe(df, target_mapping, rel="relation"):
    df_out = df.copy()
    
    # We use these specific columns just to determine IF we need to swap
    ht = df_out["x_type"].astype(str).values
    tt = df_out["y_type"].astype(str).values
    hn = df_out["x_name"].astype(str).values
    tn = df_out["y_name"].astype(str).values
    rels = df_out[rel].astype(str).values
    
    # Boolean mask tracking which rows must be entirely flipped
    swap_mask = np.zeros(len(df_out), dtype=bool)
    
    # 1. Flip based on Explicit Rule Directions
    for relation, (req_head_type, req_tail_type) in target_mapping.items():
        rel_mask = (rels == relation)
        needs_swap = rel_mask & (ht == req_tail_type) & (tt == req_head_type)
        swap_mask = swap_mask | needs_swap

    # 2. Same-Type Relations (e.g., drug_drug): Sort alphabetically by name
    same_type_mask = (ht == tt)
    alpha_swap = same_type_mask & (hn > tn)
    swap_mask = swap_mask | alpha_swap
    
    # 3. Fallback for unlisted relations connecting different types: Sort alphabetically by type
    unlisted_mask = ~(same_type_mask | np.isin(rels, list(target_mapping.keys())))
    fallback_swap = unlisted_mask & (ht > tt)
    swap_mask = swap_mask | fallback_swap
    
    # 4. SAFELY SWAP THE ENTIRE NODE PAYLOAD
    # Find all x_ columns that have a matching y_ column
    x_cols = [c for c in df_out.columns if c.startswith("x_")]
    valid_pairs = [(x_col, "y_" + x_col[2:]) for x_col in x_cols if "y_" + x_col[2:] in df_out.columns]
    
    for x_col, y_col in valid_pairs:
        # Extract the original payload columns
        x_vals = df_out[x_col].values.copy()
        y_vals = df_out[y_col].values.copy()
        
        # Overwrite the dataframe: if swap_mask is True, put Y in X, otherwise keep X
        df_out[x_col] = np.where(swap_mask, y_vals, x_vals)
        df_out[y_col] = np.where(swap_mask, x_vals, y_vals)
    
    # 5. Finally, deduplicate using the fully canonicalized names
    # Subset must include all columns we care about deduplicating on
    df_out = df_out.drop_duplicates(subset=["x_name", rel, "y_name"]).reset_index(drop=True)
    
    return df_out

# Execute safely:
final_kg = enforce_explicit_directions_safe(prime_kg_df, target_mapping=EXPLICIT_DIRECTIONS)
relation_type_stats(final_kg)
```
